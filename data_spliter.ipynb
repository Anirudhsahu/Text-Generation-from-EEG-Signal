```python
subjects = df.subject_id.unique()
print(subjects)
```

```python
test_subjects = ["YAC", "YDR"]
```

```python
train_df = df[~df.subject_id.isin(test_subjects)]
test_df  = df[df.subject_id.isin(test_subjects)]

print("Train samples:", len(train_df))
print("Test samples:", len(test_df))

train_df.to_pickle("training.pkl")
test_df.to_pickle("testing.pkl")
```

```python
import os

os.makedirs("test_samples", exist_ok=True)
```

```python
for idx, row in test_df.iterrows():
    sample = row.to_dict()
    filename = f"test_samples/sample_{idx}.pkl"
    
    pd.to_pickle(sample, filename)
```
